# -------------------- LIST OPERATIONS --------------------

```python
# Creating a list
my_list = [10, 20, 30, 40, 50]
print("Original List:", my_list)
```

```python
# Append an element
my_list.append(60)
print("After append():", my_list)
```

```python
# Insert an element
my_list.insert(2, 25)
print("After insert():", my_list)
```

```python
# Remove an element
my_list.remove(25)
print("After remove():", my_list)
```

```python
# Pop an element
removed_element = my_list.pop()
print("Element removed using pop():", removed_element)
print("After pop():", my_list)
```

```python
# Sort the list
my_list.sort(reverse=True)
print("After sort():", my_list)
```


# -------------------- TUPLE OPERATIONS --------------------

```python

# Creating a tuple
my_tuple = (10, 20, 30, 20, 40, 50)
print("\nOriginal Tuple:", my_tuple)
```

```python
# Accessing an element
print("Element at index 2:", my_tuple[2])
```

```python
# Counting an element
print("Count of 20:", my_tuple.count(20))
```

```python
# Finding the index of an element
print("Index of 40:", my_tuple.index(40))
```

```python
# Slicing
print("Tuple after slicing:", my_tuple[1:5])
```

```python
# Concatenation
new_tuple = my_tuple + (60, 70)
print("After concatenation:", new_tuple)
```

```python
# Repetition
repeated_tuple = (1, 2) * 3
print("After repetition:", repeated_tuple)
```


# -------------------- DICTIONARY OPERATIONS --------------------


```python

# Creating a dictionary
my_dict = {
    "Name": "Aayu",
    "Age": 18,
    "Course": "Computer Engineering"
}
print("\nOriginal Dictionary:", my_dict)
```

```python
# Adding a new key-value pair
my_dict["Year"] = 1
print("After adding an element:", my_dict)
```

```python
# Updating a value
my_dict["Age"] = 19
print("After updating Age:", my_dict)
```

```python
# Accessing a value using get()
print("Course:", my_dict.get("Course"))
```

```python
# Displaying keys
print("Keys:", my_dict.keys())
```

```python
# Displaying values
print("Values:", my_dict.values())
```

```python
# Displaying key-value pairs
print("Items:", my_dict.items())
```

```python
# Removing an element using pop()
removed_value = my_dict.pop("Year")
print("Removed value:", removed_value)
print("After pop():", my_dict)
```
