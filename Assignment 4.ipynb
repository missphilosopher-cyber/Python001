```python
import numpy as np

# Creating matrices using Python lists
matrix1 = [[7, 11],
           [9, 11]]

matrix2 = [[6, 7],
           [6, 9]]

print("Using Python Lists:")
print("The first matrix is", matrix1)
print("The second matrix is", matrix2)

# Matrix addition using lists
result_list = [[matrix1[i][j] + matrix2[i][j] for j in range(2)] for i in range(2)]

print("The addition of both matrices is", result_list)


# Creating NumPy arrays
array1 = np.array(matrix1)
array2 = np.array(matrix2)

print("\nUsing NumPy Arrays:")
print("The first matrix is", array1)
print("The second matrix is", array2)

# Matrix addition using NumPy
result_array = array1 + array2

print("The addition of both matrices is", result_array)
```
