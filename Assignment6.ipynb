```python
import pandas as  pd
df = pd.read_csv("cleaned_Student_Academic_Performance.csv")
# Step 4: Handle missing numerical values using mean
numeric_cols = df.select_dtypes(include="number").columns

df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# Handle missing categorical values using mode
categorical_cols = df.select_dtypes(include="object").columns

if len(categorical_cols) > 0:
    df[categorical_cols] = df[categorical_cols].fillna(
        df[categorical_cols].mode().iloc[0]
    )

print("\nDataset After Handling Missing Values:")
print(df)
```

```python
# Step 5: Remove remaining missing records
df = df.dropna()
```

```python
# Step 6: Identify and remove duplicate records
print("\nDuplicate Records:")
print(df.duplicated())

df = df.drop_duplicates()
```

```python

# Step 7: Remove unnecessary column if present
if "Student_ID" in df.columns:
    df = df.drop(columns=["Student_ID"])
else:
    print("\nNo Student_ID column to remove.")
```

```python

# Step 8: Select particular columns using loc[]
print("\nSelected Columns:")
print(df.loc[:, [df.columns[0], df.columns[1]]])


# Filter records using a numerical column
if len(numeric_cols) > 0:
    column_name = numeric_cols[0]

    average_value = df[column_name].mean()

    result = df.loc[df[column_name] >= average_value]

    print("\nFiltered Records:")
    print(result)
```

```python

# Step 9: Display final preprocessed dataset
print("\nFinal Preprocessed Dataset:")
print(df)

print("\nFinal Dataset Information:")
df.info()

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())
```
